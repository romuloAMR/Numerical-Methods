# Numerical integrations

Implementation of integrations using numerical computational methods 💻

## Definition:

Numerical integration is the approximate calculation of an integral using computational methods when the exact solution is difficult or impossible to obtain.

## All imports

```python
import numpy as np
from typing import Callable
import sys
```

## Overall parameters

```python
start_range = 1
end_range = 3
subranges_number = 6
f: Callable[[float], float] = lambda x: 1/x
```

## Left Riemann Sum:

```python
def left_riemann_sum(
        f: Callable[[float], float],
        start_range: float,
        end_range: float,
        subranges_number: int | None = None 
    ) -> float:

    if end_range < start_range:
        start_range, end_range = end_range, start_range

    if subranges_number is None or subranges_number < 1:
        subranges_number = int((end_range-start_range)*100)

    width = (end_range-start_range)/float(subranges_number)

    return width * sum(f(start_range + width * i) for i in range(subranges_number))

left_riemann_sum(f, start_range, end_range, subranges_number)
```

## Right Riemann Sum:

```python
def right_riemann_sum(
        f: Callable[[float], float],
        start_range: float,
        end_range: float,
        subranges_number: int | None = None 
    ) -> float:

    if end_range < start_range:
        start_range, end_range = end_range, start_range

    if subranges_number is None or subranges_number < 1:
        subranges_number = int((end_range-start_range)*100)

    width = (end_range-start_range)/float(subranges_number)

    return width * sum(f(start_range + width * i) for i in range(1, subranges_number+1))

right_riemann_sum(f, start_range, end_range, subranges_number)
```
